```python
# 2.	Scenario: A program processes a list of numbers and needs to find the largest value.
 # Write logic to identify and return the largest number from a given list.
def largest_number(numbers):
    largest=numbers[0]
    for num in numbers:
        if(num>largest):
            largest=num
    return largest

result1=largest_number([89,98,76,56])
print("largest number found",result1)
    
def smallest_number(numbersmall):
    smallest=numbersmall[0]
    for num in numbersmall:
        if(num<smallest):
            smallest=num
    return smallest
result2=smallest_number([1,-23,76,89,87])
print(result2)
```

```python
# 7.	Scenario: A retail store offers a 20% discount if a customer’s total order exceeds $100. 
# Write logic to calculate the final amount to be paid after applying the discount.

# Total order >= 100
# dicount = total order *(20/100)
# final amount = Total order - discount

def offer_discount(Total_order):
    if(Total_order>=100):
        dicount = Total_order *(20/100)
        final_amount = Total_order - dicount
        print("Final Amount after discount :",final_amount)
        offer_price='offer applied successfully'

    else:
        print("Discount not applicable for the purchased amount")
        offer_price='offer not applied'
    return offer_price

offer_discount(10)    
        
    
    
```

```python
# 8.	Scenario: A banking system processes withdrawal requests and ensures the user has enough balance.

 # Write logic to check if a user has enough balance before allowing a withdrawal and update the remaining balance accordingly.

# required amount = 500
# balance amount =1000
# if ba >= ra -> withdrawal processed
# remaining balance = balance amount - required amount 

def Withdrawal_process():
    Balance_Amount=int(input("Enter the balance amount:"))
    Withdrawal_Amount=int(input("Enter the withdrawal amount:"))
    if (Balance_Amount>=Withdrawal_Amount):
        print("Withdrawal can be processed")
        Remaining_Balance=Balance_Amount-Withdrawal_Amount
        print("Balance fetched :",Remaining_Balance)
        Customer_status='Good civil score'

    else:
        print("Low balance withdrawal cannot be processed")
        Customer_status='Moderate civil score'
    return Customer_status

Withdrawal_process()
    

        
```

```python
# Scenario: A calendar system verifies whether a given year is a leap year based on standard leap year rules.
 # Write logic to determine whether a given year is a leap year.
# year/4 . year/400-> leap year 
# year/100 -> not leap year

Year= int(input("Enter the year "))
if(Year%400==0):
    print(Year,"Given year is Leap year")
elif(Year%100==0):
    print(Year,"Given year is not a Leap year")
elif(Year%4==0):
    print(Year,"Given year is a Leap year")
else:
    print(Year,"Given year is a Leap year")
     
      
     
      
      
              
```

```python
# 10.	Scenario: A program filters out only even numbers from a given list.
 # Write logic to extract and return only the even numbers from a list.
# num%2==0 -> even number 
# empty list
# append 

def Even_numbers(numbers):
    Even_list=[]
    for num in numbers:
        if (num%2==0):
            Even_list.append(num)
    return Even_list

Even_numbers([34,76,21,33,24,79,800,65432,76542,98764,3334487])

            
    
```
