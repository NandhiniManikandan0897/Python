```python
# 3.Write a Python function to check if a given string is a palindrome.
# Expected Output (for input "madam"):
# madam is a palindrome

def check_palindrome(text):
    check_text = text[::-1]
    if text == check_text:
        print("given text is palidrome" )
    else:
        print("given text is not palidrome")
              
check_palindrome("space")    
    
```

```python
# 4.Write a Python program to count the number of vowels in a given string. 
# Expected Output (for input "education"):

# Number of vowels: 5
find_vowels = str(input("enter the word "))
vowels = ['a','e','i','o','u']
vowels_in_word = []
new_word = list(find_vowels)

for word in new_word:
    if(word in vowels):
        vowels_in_word.append(word)
        Number_of_vowels = len(vowels_in_word)
print("Number of vowels in word",Number_of_vowels)
     
    



     
```

```python
# 5.Write a Python program to remove duplicates from a list.
# Expected Output (for input [1, 2, 2, 3, 4, 4, 5]):

#  [1, 2, 3, 4, 5]
# n = n+1 -->remove

given_list=[1, 2, 2, 3, 4, 4, 5]
new_list=[]
for n in given_list:
    if n not in new_list:
        new_list.append(n)
print(new_list)        
        
        
```
